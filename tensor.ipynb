## Flatten()

```python
import torch
```

```python
x = torch.arange(24)
x = x.reshape(4,6)
print(x.shape)

flat = x.flatten()
print(flat)
print(flat.shape)
```

```python
# Flatten a part

x = torch.arange(24)
x = x.reshape(2,3,2,2)
print(x)

x = x.flatten(start_dim=1)
x
```

```python
x = torch.arange(24).reshape(2,3,4)

print(x.flatten())
print(x.flatten(start_dim=1))
print(x.reshape(-1))
```
